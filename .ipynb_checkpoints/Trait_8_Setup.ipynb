```python
df = df.sort_values(by='class')
df
```

```python
#grouping by individual class:

groups = df.groupby('class')

for class_name, group in groups:
    df['filename'] = -1*df['filename']
    print(class_name)
    print(group)
```

```python
df.to_csv("file_organized_by_Class.csv")
```

```python
class_dfs = {
    name: group.reset_index(drop=True)
    for name, group in df.groupby('class')
}

print(class_dfs)
```

```python
#Potential plots

import matplotlib.pyplot as plt

colors= {
    'Crown': 'black',
    'protocone': 'red',
    'Hypocone': 'blue',
    'Paracone': 'green',
    'Metacone': 'orange'
}

for class_name, group in df.groupby('class'):
    plt.scatter(group['xmin'], group['ymin'], color=colors.get(class_name, 'gray'),
    label=class_name)

plt.xlabel('X Coords(in pixels)')
plt.ylabel('Y Coords (in pixels)')
plt.legend()
plt.grid()
plt.axis('equal')
plt.title("tooth Morphology by Class")


plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

colors = {
    'Crown': 'black',
    'Protocone': 'red',
    'Hypocone': 'blue',
    'Paracone': 'green',
    'metacone': 'orange'
}

fig, ax = plt.subplots()

for _, row in df.iterrows():
    rect = patches.Rectangle(
        (row['xmin'], row['ymin']),
        row['xmax'] - row['xmin'],
        row['ymax'] - row['ymin'],
        linewidth=2,
        edgecolor=colors.get(row['class'], 'gray'),
        facecolor='none'
    )
    ax.add_patch(rect)

ax.set_xlabel('X Coordinate (pixels)')
ax.set_ylabel('Y Coordinate (pixels)')
ax.set_title('Tooth Morphology by Class')
ax.invert_yaxis()
ax.set_aspect('equal')
plt.show()
```
